```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import ComplementNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
iris = load_iris()
X = iris.data
Y = iris.target

print(X)
print(Y)
```

```python
iris
```

```python
iris.target_names
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=42)
```

```python
model = ComplementNB()
model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)

accuracy = accuracy_score(Y_test, y_pred)
report = classification_report(Y_test, y_pred, target_names=iris.target_names)

print(f'Rocnost modela: {accuracy:.3f}')
print(f'Izvjestaj klasifikacije:\n{report}')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report

pd.set_option('display.float_format', '{:,.2f}'.format)
```

```python
wine = load_wine()
wine.keys()
```

```python
print(wine.DESCR)
```

```python
df = pd.DataFrame(wine.data, columns=wine.feature_names)
df
```

```python
wine.target_names
```

```python
df['target'] = wine.target

df['wine_class'] = df['target'].map({
    0: wine.target_names[0],
    1: wine.target_names[1],
    2: wine.target_names[2]
})

df.head(10)
```

```python
df.shape
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.describe().T
```

```python
df.groupby('wine_class')[wine.feature_names].mean().T
```

```python
for wine_class in df['wine_class'].unique():
    subset = df[df['wine_class'] == wine_class]
    plt.scatter(
        subset['alcohol'],
        subset['flavanoids'],
        label=wine_class
    )

plt.title('Uzorci vina prema alkoholu i flavanoidima')
plt.xlabel('Alkohol')
plt.ylabel('Flavanoidi')
plt.legend()
plt.grid(True)
plt.show()
```

X - ulazne znacajke

y - ciljane varijable


X - vrijednosti kemijskih mjerenja

y - klase vina

```python
X = df[wine.feature_names]
y = df['target']
```

```python
X.head()
```

```python
y.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
```

```python
X_train.shape, X_test.shape
```

```python
tree_model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

tree_model.fit(X_train, y_train)
```

```python
y_pred_tree = tree_model.predict(X_test)
y_pred_tree
```

```python
resultati_tree = X_test.copy()

resultati_tree['data_klasa'] = y_test
resultati_tree['pred_klasa'] = y_pred_tree

resultati_tree['data_naziv_klasa'] = resultati_tree['data_klasa'].map({
    0: wine.target_names[0],
    1: wine.target_names[1],
    2: wine.target_names[2]
})

resultati_tree['pred_naziv_klasa'] = resultati_tree['pred_klasa'].map({
    0: wine.target_names[0],
    1: wine.target_names[1],
    2: wine.target_names[2]
})

resultati_tree[['data_naziv_klasa', 'pred_naziv_klasa']].head(10)
```

```python
accuracy_tree = accuracy_score(y_test, y_pred_tree)
accuracy_tree
```

KNN - K-Nearest Neighbors

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
```

```python
y_pred_knn = knn_model.predict(X_test)
y_pred_knn
```

```python
resultati_knn = X_test.copy()

resultati_knn['data_klasa'] = y_test
resultati_knn['pred_klasa'] = y_pred_knn

resultati_knn['data_naziv_klasa'] = resultati_knn['data_klasa'].map({
    0: wine.target_names[0],
    1: wine.target_names[1],
    2: wine.target_names[2]
})

resultati_knn['pred_naziv_klasa'] = resultati_knn['pred_klasa'].map({
    0: wine.target_names[0],
    1: wine.target_names[1],
    2: wine.target_names[2]
})

resultati_knn[['data_naziv_klasa', 'pred_naziv_klasa']].head(10)
```

```python
accuracy_knn = accuracy_score(y_test, y_pred_knn)
accuracy_knn
```

```python
usporedba = pd.DataFrame({
    'model': ['Decision Tree', 'K-Neighbors'],
    'accuracy': [accuracy_tree, accuracy_knn]
})

usporedba
```
